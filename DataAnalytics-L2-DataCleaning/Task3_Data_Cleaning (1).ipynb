```python
# TASK 3 - DATA CLEANING
# OASIS Infobyte Internship

import pandas as pd
import numpy as np

# Load the dataset
file_path = "/content/Messy_Employee_dataset.csv"
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

# Display first 5 rows
display(df.head())
```

```python
# Data Quality Report

print("===== DATA QUALITY REPORT =====\n")

# 1. Dataset shape
print("Dataset shape:", df.shape)

# 2. Column names
print("\nColumn names:")
print(df.columns.tolist())

# 3. Missing values
print("\nMissing values per column:")
print(df.isnull().sum())

# 4. Duplicate rows
print("\nNumber of duplicate rows:", df.duplicated().sum())

# 5. Data types
print("\nData types:")
print(df.dtypes)

# 6. Unique values per column
print("\nUnique values per column:")
print(df.nunique())

# 7. Numerical summary
print("\nNumerical columns summary:")
display(df.describe().T)
```

```python
### 1. Dataset Size - The dataset contains 1,020 rows and 12 columns.

### 2. Missing Values - Missing values are present, particularly in `Age`, `Salary`, and `Performance_Score`.

### 3. Employee ID - Employee_ID` contains 1,020 unique values, indicating that it can be used as a unique employee identifier.

### 4. Join Date - The `Join_Date` column contains date values and requires date-format standardisation.

### 5. Performance Score - Performance_Score` is stored as an `object` data type and requires data-type checking and correction.

### 6. Phone - The `Phone` column contains numeric values but should be treated as an identifier/string rather than a numerical measurement.

### 7. Overall Data Quality - The dataset requires further cleaning to handle missing values, inconsistent data types/formats, duplicates, and possible numerical anomalies.
```

```python
# Missing Data Analysis

print("Missing values before cleaning:")
missing_values = df.isnull().sum()

missing_summary = pd.DataFrame({
    'Missing Count': missing_values,
    'Missing Percentage': (missing_values / len(df) * 100).round(2)
})

display(missing_summary[missing_summary['Missing Count'] > 0])
```

```python
### 1. Age Missing Values - The `Age` column contains 211 missing values, representing 20.69% of the dataset.

### 2. Salary Missing Values - The `Salary` column contains 24 missing values, representing 2.35% of the dataset.

### 3. Missing Data Handling Strategy - Missing values in `Age` will be replaced using the median age because the median is less sensitive to extreme values.

### Missing values in `Salary` will also be replaced using the median salary to preserve the existing rows and avoid unnecessary data loss.

### 4. Cleaning Decision - Median imputation is selected for both numerical columns because it provides a simple and robust method for handling the missing observations while maintaining the size of the dataset.
```

```python
# Step 2: Handle Missing Values

# Fill missing Age values with the median Age
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill missing Salary values with the median Salary
df['Salary'] = df['Salary'].fillna(df['Salary'].median())

# Check missing values after imputation
print("Missing values after cleaning:")

missing_after = df.isnull().sum()

display(
    missing_after[missing_after > 0].to_frame('Missing Count')
)
```

```python
### Missing Data Handling - Observation

### Missing values were identified in the `Age` and `Salary` columns.
### The `Age` column had 211 missing values, representing 20.69% of the dataset.
### The `Salary` column had 24 missing values, representing 2.35% of the dataset.
### Median imputation was used for both numerical columns.
### The median was selected because it is less affected by extreme values than the mean.
### No rows were deleted, so the original dataset size was preserved.
```

```python
# Step 3: Duplicate Removal

# Count duplicate rows before removal
duplicate_count = df.duplicated().sum()

print("Duplicate rows before cleaning:", duplicate_count)

# Remove duplicate rows
df = df.drop_duplicates().reset_index(drop=True)

# Display dataset shape after removing duplicates
print("Dataset shape after duplicate removal:", df.shape)
```

```python
### 3. Duplicate Removal - Observation

### Duplicate Check
### The dataset contains **0 duplicate rows**.
### No duplicate records were removed.
### The dataset remains **1,020 rows and 12 columns** after duplicate removal.
### The original row count was preserved because no duplicates were found.
```

```python
# Step 4: Standardization - Inspect Categorical and Date Values

print("Unique values in categorical columns:")
print("\nFirst_Name:")
print(df['First_Name'].unique())

print("\nLast_Name:")
print(df['Last_Name'].unique())

print("\nDepartment_Region:")
print(df['Department_Region'].unique())

print("\nStatus:")
print(df['Status'].unique())

print("\nPerformance_Score:")
print(df['Performance_Score'].unique())

print("\nJoin_Date sample:")
print(df['Join_Date'].head(10))
```

```python
# Step 4: Standardization

# Remove leading/trailing whitespace from text columns
text_columns = df.select_dtypes(include='object').columns

for col in text_columns:
    df[col] = df[col].str.strip()

# Convert Join_Date from object/string to datetime
df['Join_Date'] = pd.to_datetime(df['Join_Date'], errors='coerce')

# Display updated data types
print("Data types after standardization:")
print(df.dtypes)

# Display sample of standardized dates
print("\nStandardized Join_Date:")
print(df['Join_Date'].head())
```

```python
### 4. Standardization - Observation

### Date Standardization
### The `Join_Date` column was converted from object/string format to datetime format.
### Date conversion ensures that the column can be used correctly for date-based analysis.
### Invalid date values, if any, were converted to missing values using `errors='coerce'`.

### Text Standardization
### Leading and trailing whitespace was removed from text columns.
### Existing categorical values such as `Status` and `Performance_Score` were already consistently formatted.
### No unnecessary changes were made to valid categorical values.
```

```python
# Step 5: Outlier Detection using IQR

# Select numerical columns for outlier detection
numeric_columns = ['Age', 'Salary']

outlier_summary = {}

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    outlier_summary[col] = len(outliers)

    print(f"{col}:")
    print(f"  Q1 = {Q1:.2f}")
    print(f"  Q3 = {Q3:.2f}")
    print(f"  IQR = {IQR:.2f}")
    print(f"  Lower Bound = {lower_bound:.2f}")
    print(f"  Upper Bound = {upper_bound:.2f}")
    print(f"  Number of Outliers = {len(outliers)}")
    print()
```

```python
### 5. Outlier Detection - Observation

### Outlier Analysis
### The IQR method was used to detect outliers in the numerical columns `Age` and `Salary`.
### The `Age` column has **0 outliers** based on the calculated IQR bounds.
### The `Salary` column has **0 outliers** based on the calculated IQR bounds.
### Since no outliers were detected, no rows were removed and no values were capped.
### The numerical data was therefore retained without outlier treatment.
```

```python
# Step 6: Data Type Correction

# Employee ID should be treated as a string
df['Employee_ID'] = df['Employee_ID'].astype(str)

# Phone should be treated as a string/identifier
df['Phone'] = df['Phone'].astype(str)

# Salary should be a numeric floating-point value
df['Salary'] = pd.to_numeric(df['Salary'], errors='coerce').astype(float)

# Age should be numeric
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

# Join_Date should remain datetime
df['Join_Date'] = pd.to_datetime(df['Join_Date'], errors='coerce')

# Display final data types
print("Data types after correction:")
print(df.dtypes)
```

```python
### 6. Data Type Correction - Observation

### Data Type Observation
###`Employee_ID` was converted to a string/object type because it is an identifier.
### `Phone` was converted to a string/object type because phone numbers should not be treated as numerical measurements.
### `Age` was confirmed as a numeric column with `float64` data type.
### `Salary` was converted to `float64` because it represents monetary values.
### `Join_Date` was converted to `datetime64[ns]` to support proper date-based analysis.
### `Remote_Work` remains a Boolean column because it represents True/False values.
### The dataset now has appropriate data types for further analysis.
```
